## Linear Algebra

These are my notes from an intro linear algebra course.


### Core ideas

- **Span:** the set of all linear combinations of a set of vectors.
- **Linear transformation:** a rule that takes in a vector and outputs another vector.

A transformation is linear if:

1. all lines stay lines (no curves), and
2. the origin stays fixed.

A linear transformation also keeps grid lines parallel and evenly spaced.






### Pivots and independence

- If every column has a pivot, the vectors are linearly independent.
- If at least one column has no pivot, the vectors are linearly dependent.




### Span, independence, basis, and rank

- **Spanning set:** can reach every vector in the target space.
- **Linearly independent:** no vector in the set is redundant.
- **Basis:** a set that is both spanning and linearly independent.

To be linearly independent, you need a pivot in every column.

To span $\mathbb{R}^n$, you need a pivot in every row.

To be a basis, you need both.

**Rank** is the number of pivots in the matrix after row reduction.

Equivalent meanings:

- number of linearly independent rows,
- number of linearly independent columns,
- dimension of the space spanned by the columns.

For a matrix $A \in \mathbb{R}^{m \times n}$, the number of linearly independent columns equals the number of linearly independent rows. This value is the rank, denoted by $\mathrm{rk}(A)$.



### Gaussian elimination (row operations)

1. Swap two rows.
2. Multiply a row by a nonzero number.
3. Add or subtract a multiple of one row to another row.

### Linear mapping (transformation)

For vector spaces $V$ and $W$, a mapping $\Phi: V \to W$ is linear if for all $x, y \in V$ and scalars $\lambda, \psi \in \mathbb{R}$:

$$
\Phi(\lambda x + \psi y) = \lambda\Phi(x) + \psi\Phi(y)
$$

### Kernel intuition

The kernel tells you which directions disappear under the transformation.
