## What is a hypothesis test?

A hypothesis test uses a sample of data in an experiment to test a statement made about the value of a **population parameter**.

- It is used when the value of the assumed population parameter is **questioned**
- It looks at which outcomes are **unlikely** to occur if the assumed population parameter is true
- The probability found is compared against a given **significance level** to determine whether there is evidence to believe that the assumed population parameter is not true

---

## Key Terms

| Term | Definition |
|------|------------|
| **Hypothesis** | An assumption made about a particular population parameter |
| **Population parameter** | A numerical characteristic which helps define a population (e.g. probability *p* of an event, or the mean of a population) |
| **Null hypothesis (H₀)** | The assumed population parameter given that no change has happened |
| **Alternative hypothesis (H₁)** | How we think the population parameter could have changed |
| **Test statistic** | A numerical value calculated from the data (also called the *observed value*) |
| **Sample** | A subset of data taken from the population |
| **Significance level** | The smallest probability at which an event could have occurred by chance |

---

## Null & Alternative Hypotheses

Every hypothesis test must begin with a clear **null hypothesis** (what we already believe to be true) and **alternative hypothesis** (how we believe the data might have changed).

- When a hypothesis test is carried out, the null hypothesis is **assumed to be true** and will either be accepted or rejected

---

## One-Tailed vs Two-Tailed Tests

| Type | Null Hypothesis | Alternative Hypothesis | Use when... |
|------|----------------|----------------------|-------------|
| **One-tailed** | H₀ : θ = ... | H₁ : θ > ... **or** H₁ : θ < ... | Testing if the parameter has **increased or decreased** |
| **Two-tailed** | H₀ : θ = ... | H₁ : θ ≠ ... | Testing if the parameter has **changed** in either direction |

> ⚠️ Read the wording of the question carefully to decide whether your test should be one-tailed or two-tailed.

---

## Significance Level

- Sets the smallest probability that an event could have occurred **by chance**
- Any probability **smaller** than the significance level suggests the event is unlikely to have happened by chance
- Must be set **before** the hypothesis test is carried out
- Common values: **1%**, **5%**, or **10%**

## Rejecting or Accepting H₀

The null hypothesis is **rejected** if:
- The **p-value** is less than the significance level, **or**
- The **test statistic** falls within the **critical region**

---

### P-Value

The p-value is the probability of observing a value **at least as extreme** as the test statistic, assuming H₀ is true.

| Test Direction | What counts as "extreme"? | Find... |
|---------------|--------------------------|---------|
| **Decrease** | Values smaller than the test statistic | P(X ≤ test statistic) |
| **Increase** | Values larger than the test statistic | P(X ≥ test statistic) |
| **Two-tailed** | Values in either direction | **Double** the one-tail probability |

> 💡 For a two-tailed test, it's common to **halve the significance level** and compare it with the one-tail probability (rather than doubling the probability).

---

### Critical Region & Critical Value

| Term | Definition |
|------|------------|
| **Critical region** | The range of test statistic values that lead to **rejecting H₀** |
| **Critical value** | The **boundary** of the critical region — the least extreme value that still leads to rejection |
| **Actual significance level** | The true probability of incorrectly rejecting H₀ (relevant for discrete distributions) |

- The critical value is determined by the **significance level**
- In a **two-tailed test**, the significance level is halved and **both tails** are tested
- For **discrete distributions**, the critical value is the first value inside the critical region, so the actual significance level may be **lower** than the stated level

---

### Critical Regions by Test Type

For an **α%** significance level:

| Test Type | Critical Region |
|-----------|----------------|
| **One-tailed** | **α%** in the tail being tested |
| **Two-tailed** | **α/2 %** in **each** tail |

#### We’ll use the following function to compute the absolute difference between the observed number and the expected number if the coin is fair.

```python
import numpy as np

n = 250
p = 0.5


def abs_deviation(heads):
    expected = n * p
    return np.abs(heads - expected)
```
